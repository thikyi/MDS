In the cell below, define a function named `first_and_last_sum` which:

- Takes a list of numbers
- Returns the sum of the first and last elements of the list, as long as the list contains at least two elements

If the input list contains fewer than two elements, then `first_and_last_sum` should throw a `NotEnoughElementsError` exception.
Class `NotEnoughElementsError` will also need to be defined.
`NotEnoughElementsError` objects contain a field `count`, which will hold the number of elements the list actually contained.
Example calls are provided in the next cell; leave these in place in order to test your code.

**For full credit, your solution must use a negative list index somewhere.**

````python
# Define your NotEnoughElementsError exception and your
# first_and_last_sum function below.  Leave the calls in place
# to test your code.
class NotEnoughElementsError(Exception):
    def __init__(self, count):
        self.count = count

def first_and_last_sum(li):
   if len(li) >= 2:
       return li[0] + li[-1]
   else:
       raise NotEnoughElementsError(len(li))
    
#-----------------------------------------------------


try:
    print(f"Sum: {first_and_last_sum([])}")
except NotEnoughElementsError as err:
    print(f"Error; count: {err.count}")
# Above try...except should have the following line of output:
# Error; count: 0

try:
    print(f"Sum: {first_and_last_sum([5])}")
except NotEnoughElementsError as err:
    print(f"Error; count: {err.count}")
# Above try...except should have the following line of output:
# Error; count: 1

try:
    print(f"Sum: {first_and_last_sum([5, 2])}")
except NotEnoughElementsError as err:
    print(f"Error; count: {err.count}")
# Above try...except should have the following line of output:
# Sum: 7

try:
    print(f"Sum: {first_and_last_sum([5, 3, 1])}")
except NotEnoughElementsError as err:
    print(f"Error; count: {err.count}")
# Above try...except should have the following line of output:
# Sum: 6

try:
    print(f"Sum: {first_and_last_sum([4, 2, 1, 7, 8])}")
except NotEnoughElementsError as err:
    print(f"Error; count: {err.count}")
# Above try...except should have the following line of output:
# Sum: 12
````

Consider the following cell, which defines a function named `average_greater_than`.
`average_greater_than` takes the following parameters:

1. A list of numbers
2. A minimum value (`minimum`), a number

Given these parameters, `average_greater_than` will compute the average of all numbers in the list which are greater than or equal to `minimum`.

````python
def average_greater_than(numbers, minimum):
    total = 0
    count = 0
    for n in numbers:
        if n >= minimum:
            total += n
            count += 1
    return total / count

print(average_greater_than([5, 2, 4, 1, 3], 3)) # prints 4.0 - the average of 5, 4, and 3
print(average_greater_than([3, 7, 2, 10, 8, 12], 4)) # prints 9.25 - the average of 7, 10, 8, and 12
print(average_greater_than([3, 7, 2, 10, 8, 12], 9)) # prints 11.0 - the average of 10 and 12
````
